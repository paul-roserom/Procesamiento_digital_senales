# FFT Y FILTRO NOTCH

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.fftpack as fft
import cv2
from numpy import fft
```

```python
archivo = '/home/paul/Documentos/Personal/UIDE/MATERIAS/Semana 2/PRACTICAS/Data/img2.png'
img = cv2.imread(archivo,0) # escala de grises
rows, cols = img.shape
print(rows, cols)

img_fft = fft.fftshift(fft.fft2(img))
magnitud = np.abs(img_fft)

# convertirlo en escala logaritmica
magnitud_log = 20 * np.log(1+magnitud)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(magnitud_log, cmap='magma')
plt.axis('off')
plt.show()
```

```python
busqueda_maximos = magnitud.copy()
radio_exclusion = 10 # px
centro_y, centro_x = rows // 2, cols // 2
cv2.circle(busqueda_maximos, (centro_x, centro_y), radio_exclusion, 0, -1)


# encontrar las cordenadas de interferencia
y_ruido, x_ruido = np.unravel_index(np.argmax(busqueda_maximos), busqueda_maximos.shape)

# calcular las distancias de frecuencias hacia el centro
dist_x = x_ruido - centro_x
dist_y = y_ruido - centro_y

# crear máscara del filtro
mask = np.ones((rows, cols), dtype= np.float32)

#Dibujar el ruido detectado

cv2.circle(mask,(x_ruido,y_ruido), radio_exclusion, 0, -1)

# dibujamos un parche para Fourier
cv2.circle(mask,(centro_x-dist_x,centro_y-dist_y), radio_exclusion, 0, -1)

# aplicar el filtro y regresar info a rgb

f_filtrado = img_fft * mask
img_limpia = np.abs(fft.ifft2(fft.ifftshift(f_filtrado)))

plt.imshow(img, cmap = 'gray')
plt.show()
plt.imshow(img_limpia, cmap = 'gray')
plt.show()
```

```python
import cv2
import numpy as np
from numpy import fft
import matplotlib.pyplot as plt
import os

# --- Configuración de rutas ---
path_data = '/home/paul/Documentos/FFT/Data/' # cambiar  a una ruta propia
if not os.path.exists(path_data):
    os.makedirs(path_data)

# 1. Configuración
size = 512

# 2. Captura de la foto
cam = cv2.VideoCapture(0)
print("Iniciando cámara...")
for i in range(10): 
    ret, frame = cam.read()

if not ret:
    print("Error: No se pudo capturar la foto.")
    cam.release()
else:
    print("¡Foto capturada!")
    cam.release() 

 
    # --- 3. Procesamiento ---
    image = cv2.resize(frame, (size, size))
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # FFT
    Fsh_Image = fft.fftshift(fft.fft2(np.float64(gray_image)))
    magnitud = np.abs(Fsh_Image)

    # --- 4. DETECCIÓN Y FILTRO NOTCH AUTOMÁTICO ---
    busqueda = magnitud.copy()
    centro = size // 2
    cv2.circle(busqueda, (centro, centro), 25, 0, -1)
    
    y_ruido, x_ruido = np.unravel_index(np.argmax(busqueda), busqueda.shape)
    dist_x, dist_y = x_ruido - centro, y_ruido - centro

    mask_notch = np.ones((size, size), dtype=np.float32)
    radio_notch = 15
    cv2.circle(mask_notch, (x_ruido, y_ruido), radio_notch, 0, -1)
    cv2.circle(mask_notch, (centro - dist_x, centro - dist_y), radio_notch, 0, -1)

    # --- 5. APLICACIÓN DE FILTROS ---
    FFt_filtered = Fsh_Image * mask_notch 
    
    # Inversa
    ImageFiltered = np.abs(fft.ifft2(fft.ifftshift(FFt_filtered)))
    ImageFiltered = cv2.normalize(ImageFiltered, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    # --- 6. GUARDAR IMÁGENES (.png) ---
    # Guardamos las matrices de imagen directamente para máxima calidad
    cv2.imwrite(os.path.join(path_data, 'original_camara.png'), gray_image)
    cv2.imwrite(os.path.join(path_data, 'original_camara_filter.png'), ImageFiltered)
    print(f"Imágenes guardadas en: {path_data}")

    # --- 7. VISUALIZACIÓN ---
    plt.figure(figsize=(18, 10))
    
    plt.subplot(2, 3, 1)
    plt.imshow(gray_image, cmap='gray')
    plt.title("1. Original")
    plt.axis('off')

    plt.subplot(2, 3, 2)
    plt.imshow(np.log(1 + np.abs(FFt_filtered)), cmap='magma')
    plt.title("2. Espectro con Notch")
    plt.axis('off')
    
    plt.subplot(2, 3, 3)
    plt.imshow(ImageFiltered, cmap='gray')
    plt.title("3. Resultado Final")
    plt.axis('off')

    plt.subplot(2, 3, 4)
    plt.hist(gray_image.ravel(), 256, [0, 256], color='gray')
    plt.title("Hist: Original")

    plt.subplot(2, 3, 6)
    plt.hist(ImageFiltered.ravel(), 256, [0, 256], color='blue')
    plt.title("Hist: Filtrado")

    plt.tight_layout()
    plt.show()
    
```

# Filtro gaussiano

```python
from skimage import io, filters, img_as_float, img_as_ubyte

imagen_original = io.imread('/home/paul/Documentos/Personal/UIDE/MATERIAS/Semana 2/PRACTICAS/Data/ir.tif')

# convertirla en float
imagen_float = img_as_float(imagen_original) 

# aplicar el filtro 
imagen_suavizada = filters.gaussian(imagen_float, sigma=0.5, channel_axis= None)  # RGB channel_axis = 1

fig, ax = plt.subplots(1,2, figsize = (12,6))

ax[0].imshow(imagen_original, cmap = 'gray')
ax[0].axis('off')

ax[1].imshow(imagen_suavizada, cmap = 'gray')
ax[1].axis('off')
plt.show()

imagen_final = img_as_ubyte(imagen_suavizada)
io.imsave('/home/paul/Documentos/Personal/UIDE/MATERIAS/Semana 2/PRACTICAS/Data/ir_suavizada.tif', imagen_final)
```

```python
import cv2
import matplotlib.pyplot as plt
from skimage import filters, feature, img_as_float
from scipy import ndimage

# 1. Cargar la imagen con OpenCV
# Usamos cv2.IMREAD_UNCHANGED para mantener la profundidad de bits de los .tiff
img_bgr = cv2.imread('/home/paul/Documentos/Personal/UIDE/MATERIAS/Semana 2/PRACTICAS/Data/satelite_1.png')

if img_bgr is None:
    print("Error: No se pudo cargar la imagen. Revisa la ruta.")
else:
    # 2. Convertir de BGR a RGB (para visualización correcta en matplotlib)
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    
    # 3. Convertir a escala de grises para los filtros de bordes
    img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    
    # 4. Normalizar a float (0 a 1)
    img_float = img_as_float(img_gray)

    # --- Aplicación de Filtros ---
    # Gaussiano (Suavizado)
    gauss = filters.gaussian(img_float, sigma=1.5)
    
    # Laplaciano (vía SciPy)
    laplace = ndimage.laplace(gauss)
    
    # Sobel y Canny (vía skimage)
    sobel = filters.sobel(img_float)
    canny = feature.canny(img_float, sigma=2.0)

    # 5. Visualización
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    ax = axes.ravel()

    ax[0].imshow(img_rgb)
    ax[0].set_title('Original (Cargada con OpenCV)')
    
    ax[1].imshow(sobel, cmap='gray')
    ax[1].set_title('Filtro Sobel')
    
    ax[2].imshow(laplace, cmap='gray')
    ax[2].set_title('Laplaciano sobre Gaussiano')
    
    ax[3].imshow(canny, cmap='gray')
    ax[3].set_title('Algoritmo Canny')

    for a in ax:
        a.axis('off')

    plt.tight_layout()
    plt.show()
```
